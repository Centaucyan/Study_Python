```python
#. p165

import datetime

now_datetime = datetime.datetime.now()

print(now_datetime)
print()
print(now_datetime.year)
print(now_datetime.month)
print(now_datetime.day)
print()
print(now_datetime.hour)
print(now_datetime.minute)
print(now_datetime.second)
print(now_datetime.microsecond)
print()
print(now_datetime.weekday())  #. 0: Monday, 1: Tuesday, 2: Wednesday, 3: Thursday, 4: Friday, 5: Saturday, 6: Sunday
print(now_datetime.date())
print(now_datetime.time())
print(now_datetime.strftime("%Y-%m-%d %H:%M:%S"))
```
