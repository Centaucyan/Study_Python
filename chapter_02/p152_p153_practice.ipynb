```python
#. p152

from math import pi

r = float(input("구의 반지름을 입력하세요: "))

volume_of_the_sphere = (4/3) * pi * r**3
area_of_the_sphere = 4 * pi * r**2

print(f"구의 반지름: {r:g}")
print()
print(f"구의 부피: {volume_of_the_sphere}")
print(f"구의 겉넓이: {area_of_the_sphere}")
print()
print(f"구의 부피: {volume_of_the_sphere:.2f}")
print(f"구의 겉넓이: {area_of_the_sphere:.2f}")
```
