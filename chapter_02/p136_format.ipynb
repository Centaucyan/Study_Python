```python
#. 149

data = ["별", 2, "M", "부산"]

print("""\
이름: {}
나이: {}
성별: {}
거주지: {}
""".format(*data))  #. 전개 연산자(*)를 사용하여 리스트의 각 요소를 format() 메서드에 전달.
```
